Student-grade-manager

```python
import statistics

students = {
    "Rahul": [85, 90, 78, 92, 88],
    "Priya": [70, 65, 80, 75, 72],
    "Aman":  [95, 98, 97, 99, 96],
    "Sneha": [55, 60, 58, 62, 57],
}

def get_grade(avg):
    if avg >= 90: return 'A+'
    elif avg >= 80: return 'A'
    elif avg >= 70: return 'B'
    elif avg >= 60: return 'C'
    else: return 'F'

print(f"{'Name':<12} {'Avg':>6} {'Grade':>6} {'Status':>10}")
print("-" * 40)
for name, marks in students.items():
    avg = statistics.mean(marks)
    grade = get_grade(avg)
    status = "Pass" if avg >= 60 else "Fail"
    print(f"{name:<12} {avg:>6.1f} {grade:>6} {status:>10}")

avgs = {n: statistics.mean(m) for n, m in students.items()}
topper = max(avgs, key=avgs.get)
print(f"\n🏆 Topper: {topper} with {avgs[topper]:.1f}%")
```
